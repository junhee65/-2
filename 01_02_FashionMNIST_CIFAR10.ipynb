```python

import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras.datasets import fashion_mnist


tf.keras.utils.set_random_seed(42)
(x_train, y_train), (x_test, y_test) = fashion_mnist.load_data()
print("Train:", x_train.shape)
print("Test:", x_test.shape)

x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0
```

```python
#1-(1)
```

```python

from tensorflow.keras import Sequential
from tensorflow.keras.layers import Input, Flatten, Dense, Dropout

#DNN
dnn = Sequential([
    Input(shape=(28, 28)),
    Flatten(),
    Dense(256, activation="relu"),
    Dropout(0.3),
    Dense(128, activation="relu"),
    Dense(10, activation="softmax")
])

#컴파일
dnn.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

dnn.summary()
```

```python

#DNN 학습
history_dnn = dnn.fit(
    x_train, y_train,
    epochs=15,
    batch_size=128,
    validation_split=0.2
)

dnn_loss, dnn_acc = dnn.evaluate(
    x_test, y_test, verbose=0
)

print("DNN Test Loss:", dnn_loss)
print("DNN Test Accuracy:", dnn_acc)
```

테스트 정확도는 88.12%이고 손실값은 0.3395로 나타난다.DNN 모형이 테스트 이미지의 약 88.12%를 올바르게 분류함음을 뜻한다.

```python
#1-(2)
```

```python
x_train_cnn = x_train.reshape(-1, 28, 28, 1)
x_test_cnn = x_test.reshape(-1, 28, 28, 1)

print("CNN Train:", x_train_cnn.shape)
print("CNN Test:", x_test_cnn.shape)
```

```python

from tensorflow.keras import Sequential
from tensorflow.keras.layers import (
    Input, Conv2D, MaxPooling2D,
    Flatten, Dense, Dropout
)

#CNN
cnn = Sequential([
    Input(shape=(28, 28, 1)),

    Conv2D(32, (3, 3), activation="relu",
           padding="same"),
    MaxPooling2D((2, 2)),

    Conv2D(64, (3, 3), activation="relu",
           padding="same"),
    MaxPooling2D((2, 2)),

    Flatten(),
    Dense(128, activation="relu"),
    Dropout(0.3),
    Dense(10, activation="softmax")
])


cnn.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)


cnn.summary()
```

```python

#CNN
history_cnn = cnn.fit(
    x_train_cnn, y_train,
    epochs=15,
    batch_size=128,
    validation_split=0.2
)


cnn_loss, cnn_acc = cnn.evaluate(
    x_test_cnn, y_test, verbose=0
)

print("CNN Test Loss:", cnn_loss)
print("CNN Test Accuracy:", cnn_acc)
```

테스트 정확도는 91.70%이고 손실값은 0.2635로 나타난다.이는 테스트 이미지 10000개 중 약 9170개를 올바르게 분류하였음을 의미한다.

```python

import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras.datasets import cifar10


tf.keras.utils.set_random_seed(42)
(x_train, y_train), (x_test, y_test) = cifar10.load_data()
print("Train:", x_train.shape)
print("Test:", x_test.shape)


x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0
y_train = y_train.ravel()
y_test = y_test.ravel()
```

```python

from tensorflow.keras import Model, Input
from tensorflow.keras.layers import (
    Flatten, Dense, BatchNormalization,
    Activation, Add, Dropout
)

# DN Residual Block
def dense_block(x, units):
    shortcut = x

    y = Dense(units, use_bias=False)(x)
    y = BatchNormalization()(y)
    y = Activation("relu")(y)

    y = Dense(units, use_bias=False)(y)
    y = BatchNormalization()(y)

    y = Add()([shortcut, y])
    return Activation("relu")(y)


# DNN + BN + Residual
inputs = Input(shape=(32, 32, 3))
x = Flatten()(inputs)

x = Dense(256, use_bias=False)(x)
x = BatchNormalization()(x)
x = Activation("relu")(x)

x = dense_block(x, 256)
x = dense_block(x, 256)

x = Dropout(0.3)(x)
outputs = Dense(10, activation="softmax")(x)

dnn = Model(inputs, outputs)

dnn.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

dnn.summary()
```

```python

# DNN
history_dnn = dnn.fit(
    x_train, y_train,
    epochs=20,
    batch_size=128,
    validation_split=0.2
)

dnn_loss, dnn_acc = dnn.evaluate(
    x_test, y_test, verbose=0
)

print("DNN Test Loss:", dnn_loss)
print("DNN Test Accuracy:", dnn_acc)
```

```python

print("Train Accuracy:",
      history_dnn.history["accuracy"][-1])

print("Validation Accuracy:",
      history_dnn.history["val_accuracy"][-1])

print("Train Loss:",
      history_dnn.history["loss"][-1])

print("Validation Loss:",
      history_dnn.history["val_loss"][-1])
```


CIFAR-10 데이터에 Batch Normalization과 잔차 연결을 적용한 DNN 모형을 적합하였다. 테스트 정확도는 38.35% 손실값은 3.9809로 나타났다.
훈련 정확도는 88.81%였으나 검증 정확도는 37.64%로약 51.17%의 차이가 발생하였다. 이는 모형이 훈련 데이터에 과도하게 적합되어 새로운 데이터에 대한 일반화 성능이 낮아졌음을 의미한다.

```python

from tensorflow.keras import Model, Input
from tensorflow.keras.layers import (
    Conv2D, BatchNormalization, Activation,
    Add, GlobalAveragePooling2D, Dense, Dropout
)

# CNN
def conv_block(x, filters):
    shortcut = x

    y = Conv2D(filters, 3, padding="same",
               use_bias=False)(x)
    y = BatchNormalization()(y)
    y = Activation("relu")(y)

    y = Conv2D(filters, 3, padding="same",
               use_bias=False)(y)
    y = BatchNormalization()(y)

    if int(shortcut.shape[-1]) != filters:
        shortcut = Conv2D(
            filters, 1, padding="same",
            use_bias=False
        )(shortcut)
        shortcut = BatchNormalization()(shortcut)

    y = Add()([shortcut, y])
    return Activation("relu")(y)


# CNN + BN + Residual Connection
inputs = Input(shape=(32, 32, 3))

x = Conv2D(32, 3, padding="same",
           use_bias=False)(inputs)
x = BatchNormalization()(x)
x = Activation("relu")(x)

x = conv_block(x, 32)
x = conv_block(x, 64)
x = conv_block(x, 64)

x = GlobalAveragePooling2D()(x)
x = Dense(128, activation="relu")(x)
x = Dropout(0.3)(x)

outputs = Dense(10, activation="softmax")(x)

cnn = Model(inputs, outputs)

cnn.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

cnn.summary()
```

```python

# CNN
history_cnn = cnn.fit(
    x_train, y_train,
    epochs=5,
    batch_size=128,
    validation_split=0.2
)


cnn_loss, cnn_acc = cnn.evaluate(
    x_test, y_test, verbose=0
)

print("CNN Test Loss:", cnn_loss)
print("CNN Test Accuracy:", cnn_acc)
```

DNN과 CNN의 분류 성능 비교

CIFAR-10 데이터에 Batch Normalization과 ResNet 구조를 적용한 DNN과 CNN의 분류 성능을 비교한 결과, 테스트 데이터에서 DNN의 정확도는 38.35%, 손실은 3.9809였으며 CNN의 정확도는 46.76%, 손실은 1.7804로 나타났다. CNN의 분류 정확도가 DNN보다 8.41% 높았고 손실도 더 낮았다.
다만 DNN은 20 epochs, CNN은 5 epochs로 학습했으므로 학습 횟수의 차이를 고려하여 결과를 해석할 필요가 있다.
